```python
%%writefile app.py
import streamlit as st

# Judul Aplikasi
st.title("FaSHioN: Show ur Fits!")
st.write("Let's learn about fashion styles together")

st.markdown("---")

# Mengubah input angka menjadi pilihan dropdown web
pilihan = st.selectbox(
    "I pick.. (Silahkan pilih gaya fashion):",
    ["Pilih gaya...", "Casual", "Vintage", "Emo", "Streetwear", "Gyaru"]
)

# Logika menampilkan Deskripsi, Tombol, dan Gambar
if pilihan == "Casual":
    st.subheader("1. Casual Style")
    st.write("**Description:** Gaya santai yang mengutamakan kenyamanan.")
    st.write("**Fun Fact:** Sangat dipengaruhi tren athleisure.")
    st.write("**Essentials:** Kaos polos, jeans, sneakers.")
    # Link gambar dari internet
    st.image("https://images.unsplash.com/photo-1515886657613-9f3515b0c78f?w=500", caption="Casual Fit")
    if st.button("Suka dengan style Casual?"):
        st.success("Nice choice! ✨")

elif pilihan == "Vintage":
    st.subheader("2. Vintage Style")
    st.write("**Description:** Mengadaptasi tren pakaian era puluhan tahun lalu.")
    st.write("**Fun Fact:** Thrifting sangat membantu tren ini naik lagi.")
    st.write("**Essentials:** Jaket kulit oversized, kemeja retro.")
    st.image("https://images.unsplash.com/photo-1534126511673-b6899657816a?w=500", caption="Vintage Fit")
    if st.button("Suka dengan style Vintage?"):
        st.success("Klasik dan berkelas! 📜")

elif pilihan == "Emo":
    st.subheader("3. Emo Style")
    st.write("**Description:** Identik dengan warna gelap dan kaos band.")
    st.write("**Fun Fact:** Booming di pertengahan tahun 2000-an.")
    st.write("**Essentials:** Skinny jeans hitam, sabuk studded.")
    st.image("https://images.unsplash.com/photo-1607990283143-e81e7a2c93ab?w=500", caption="Emo Fit")
    if st.button("Suka dengan style Emo?"):
        st.success("Rock on! 🎸")

elif pilihan == "Streetwear":
    st.subheader("4. Streetwear Style")
    st.write("**Description:** Gaya busana khas kultur jalanan anak skate/hip-hop.")
    st.image("https://images.unsplash.com/photo-1552374196-1ab2a1c593e8?w=500", caption="Streetwear Fit")
    if st.button("Suka dengan style Streetwear?"):
        st.success("Hypebeast banget! 🔥")

elif pilihan == "Gyaru":
    st.subheader("5. Gyaru Style")
    st.write("**Description:** Subkultur Jepang yang mendobrak standar kecantikan tradisional.")
    st.image("https://images.unsplash.com/photo-1529139574466-a303027c1d8b?w=500", caption="Gyaru Fit")
    if st.button("Suka dengan style Gyaru?"):
        st.success("Super glam! 💖")


print ("A. Fashion Styles")
print ("1. Casual")
print ("2. Vintage")
print ("3. Emo")
print ("4. Streetwear")
print ("5. Gyaru")
style= int(input("I pick.. (masukkan angka pilihan): "))
if style == 1:
   print ("desc, fun fact, essentials1")
elif style == 2:
   print ("desc, fun fact, essentials2")
elif style == 3:
    print ("desc, fun fact, essentials3")
elif style == 4:
    print ("desc, fun fact, essentials4")
```
